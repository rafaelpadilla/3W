## Introduction

Welcome to the **3WToolkit Workshop 2026**! This collection of Jupyter notebooks provides a step-by-step guide to understanding and using the toolkit's features for time-series analysis, fault detection, and machine learning applications — targeting not only oil well operations with the 3W Dataset, but also machine learning tasks in general.

Throughout this workshop, we will cover the most common tasks supported by the toolkit:
